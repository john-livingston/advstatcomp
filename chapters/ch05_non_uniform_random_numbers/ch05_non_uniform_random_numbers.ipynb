## Non-Uniform Random Numbers

The fundamental problem in Monte Carlo sampling is generating random variates from an arbitrary probability distribution. While uniform random numbers from $\text{Uniform}(0, 1)$ are relatively easy to generate in practice, we often need to draw samples from non-uniform distributions. Several methods exist for transforming uniform random variates into samples from more complex distributions.

### Inverse Transform Method

The inverse transform method (or inverse probability integral transform) is one of the most straightforward approaches. If $U \sim \text{Uniform}(0, 1)$ and $F$ is the cumulative distribution function of a target distribution, then

$$X = F^{-1}(U)$$

follows the target distribution. This method works well when the inverse CDF is analytically tractable. For example, to generate exponential random variables with rate $\lambda$, we can use

$$X = -\frac{1}{\lambda} \log(U)$$

**Advantages:**
- Simple to implement when $F^{-1}$ is available
- Only requires one uniform random number per sample
- Easy to generate dependent random variates

**Disadvantages:**
- Requires computing $F^{-1}$, which may be expensive or numerically unstable
- Not all distributions have closed-form inverse CDFs

### Rejection Sampling

Rejection sampling (or acceptance-rejection method) is useful when the inverse CDF is not available. The basic idea is to sample from a simpler proposal distribution $q(x)$ that majorizes the target distribution $p(x)$, and then accept or reject samples based on a probability ratio.

Suppose we want to sample from target density $p(x)$ and we have a proposal density $q(x)$ such that $p(x) \leq M q(x)$ for some constant $M > 0$. The algorithm is:

1. Sample $X \sim q(x)$
2. Sample $U \sim \text{Uniform}(0, 1)$
3. If $U \leq \frac{p(X)}{M q(X)}$, accept $X$; otherwise reject and repeat

**Advantages:**
- Works for any target distribution where we can evaluate $p(x)$
- No need for the inverse CDF

**Disadvantages:**
- Requires finding a good proposal distribution $q(x)$
- Efficiency depends on how well $q(x)$ approximates $p(x)$
- Can be wasteful if $M$ is large (many rejections)

### Composition Methods

Composition methods are used when the target distribution can be written as a mixture:

$$p(x) = \sum_{i=1}^{k} w_i p_i(x)$$

where $w_i$ are mixture weights and $p_i(x)$ are component densities. To sample:

1. Sample component index $i$ from the discrete distribution with probabilities $w_i$
2. Sample $X \sim p_i(x)$

This is particularly useful for distributions like the normal or beta that can be expressed as mixtures of simpler distributions.
